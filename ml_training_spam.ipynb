```python
import pandas as pd

data = pd.read_csv("spam.csv", encoding="latin-1")
data = data[['v1', 'v2']]
data.columns = ['label', 'message']

data.head()
```

```python
data['label'] = data['label'].map({'ham': 0, 'spam': 1})
data.head()
```

```python
data.info()
data['label'].value_counts()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.isnull().sum()
```

Data Cleaning => Lowercase, remove punctuation, removing stopwords, removing no's, stemming

```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

ps = PorterStemmer()

def clean_text(text):
    #lowercase
    text = text.lower()
    #Punctuation
    text = ''.join([c for c in text if c not in string.punctuation])
    words = text.split()
    words = [ps.stem(word) for word in words if word not in stopwords.words('english')]
    return ' '.join(words)

data['cleaned_message'] = data['message'].apply(clean_text)
```

Outlier Handling like very short msg(like 'ok') xtrmly lng msg

```python
data['length'] = data['cleaned_message'].apply(len)
data = data[data['length'] > 10]
data.head()
```

Vectorization (txt ==> num)

```python
#by TF-IDF Vector
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=3000)
X = tfidf.fit_transform(data['cleaned_message']).toarray()
```

```python
#output/ target
Y= data['label'].values
```

Training

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

Model selection -> Naive bayes

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(x_train, y_train)
```

Accuracy cehck

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

Final step, creating udfnctn for handling new emails

```python
def predict_spam(email):
    email = clean_text(email)
    email_vector = tfidf.transform([email])
    result = model.predict(email_vector)
    return "Spam" if result[0] == 1 else "Not Spam"
predict_spam("Congratulations! You won a free iPhone")
```

```python
import pickle

# Save model
with open("spam_model.pkl", "wb") as f:
    pickle.dump(model, f)

# Save TFIDF vectorizer
with open("spam_vectorizer.pkl", "wb") as f:
    pickle.dump(tfidf, f)

print("Spam model saved successfully!")
```
